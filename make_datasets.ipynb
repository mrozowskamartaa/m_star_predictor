# Make datasets

Each dataset is compiled once here and saved as netCDF (case/time/depth preserved, metadata in `.attrs`). **Append a new cell per dataset** so this notebook is the tractable record of what exists. Lagged quantities (M_i-1, M_i+1, ...) are NOT stored -- they are derived at training time by `FeatureSelector`.


```python
from data import (
    TransientDatasetMaker, ProfileTransientDatasetMaker, MeanDatasetMaker,
    DatasetMaker,
)

data_dir = '../m_star_dataset'
```

## Transient datasets (case, time)

```python
for name in ['ePBL_paper_expanded_2283_corrected',
             'ePBL_paper_validation_2283_corrected']:
    path = TransientDatasetMaker(data_dir, name).save()
    print('saved', path)
```

## Full-profile transient datasets (case, time, depth)

```python
for name in ['ePBL_paper_2283_corrected',
             'ePBL_paper_validation_2283_corrected']:
    path = ProfileTransientDatasetMaker(data_dir, name, method='interp').save()
    print('saved', path)
```
