# Analyze predictors

Everything flows through `Prediction` objects, so a neural predictor, ePBL and persistence are the same type. `restore()` on a run gives its Prediction; comparing to a *mean* predictor is just `restore()` on that run.

To restore a **pre-refactor** run, `git checkout` the old commit -- the old code and configs live there.


```python
from experiment import restore, Registry, persistence_prediction, epbl_prediction
from data import DatasetMaker
import plots
import matplotlib.pyplot as plt
import os
```

## List available runs

```python
reg = Registry('predictors')
reg.table()
```

## Restore and plot a single predictor

```python
pred = restore(os.path.join('predictors', 'predictor_1'))
plots.loss_trajectory(pred)
plt.show()
```

```python
plots.scatter(pred, quantity='state'); plt.show()
plots.scatter(pred, quantity='tendency'); plt.show()
```

```python
import json
with open(f'{pred.config.val_dataset_path.rsplit("_transient", 1)[0]}_training_set_cases.json') as f:
    val_cases = json.load(f)
plots.case_timeseries(pred, cases=range(0, 6), features=['bl_tke'], case_meta=val_cases)
plt.show()
```

## Compare two or more predictors

```python
pred_a = restore(os.path.join('predictors', 'predictor_1'))
pred_b = restore(os.path.join('predictors', 'predictor_2'))
plots.scatter(pred_a, pred_b); plt.show()
plots.rmse_table(pred_a, pred_b)
```

## Compare to ePBL and persistence baselines

```python
val_ds = DatasetMaker.load(pred.config.val_dataset_path)
epbl = epbl_prediction(val_ds, neutral_mode='Nb')
persist = persistence_prediction(val_ds)
plots.scatter(pred, epbl, persist); plt.show()
plots.rmse_table(pred, epbl, persist)
```
